```python
# plot the sales over the years using line chart
```

```python
vg_sales0 = df.groupby(['Year','Genre'])[['NA_Sales','EU_Sales','JP_Sales','Other_Sales']].sum()                 import plotly.graph_objects as go


# Create a figure
fig = go.Figure()


# Add traces with desired names
fig.add_trace(go.Scatter(x=vg_sales0.index.get_level_values('Year').unique(), y=vg_sales0.groupby(['Year'])['NA_Sales'].sum(), mode='lines+markers', name='NA_Sales', line=dict(color='blue')))
fig.add_trace(go.Scatter(x=vg_sales0.index.get_level_values('Year').unique(), y=vg_sales0.groupby(['Year'])['EU_Sales'].sum(), mode='lines+markers', name='EU_Sales', line=dict(color='red')))
fig.add_trace(go.Scatter(x=vg_sales0.index.get_level_values('Year').unique(), y=vg_sales0.groupby(['Year'])['JP_Sales'].sum(), mode='lines+markers', name='JP_Sales', line=dict(color='green')))
fig.add_trace(go.Scatter(x=vg_sales0.index.get_level_values('Year').unique(), y=vg_sales0.groupby(['Year'])['Other_Sales'].sum(), mode='lines+markers', name='Other_Sales', line=dict(color='purple')))


# Update layout
fig.update_layout(title='Sales_YOY')
fig.update_xaxes(title='Year')
fig.update_yaxes(title='Sales_value')


# Show the figure
fig.show()
```

```python
sales_by_year = df.groupby('Year')['Global_Sales'].sum()
plt.figure(figsize=(8,5))
plt.plot(sales_by_year.index, sales_by_year.values, marker='o', linestyle='-', color='b')
plt.title("Sales Over the Years")
plt.xlabel("Year")
plt.ylabel("Total Sales")
plt.grid(True)
plt.show()
```
