```python
# The average monthly hours of a employee having 2 years of experience is 167.is't it same for the employee
# having more than 2 years experience?
```

```python
new_df = df[(df.satisfactoryLevel>=0) & (df.satisfactoryLevel<=0.50) & (df['timeSpent.company']==2) & (df.avgMonthlyHours>=130)
& (df.avgMonthlyHours<=200)]
```

```python
proposed_mean=167
```

```python
sample_df=new_df.sample(100,random_state=0)
```

```python
sample_df
```

```python
# t-test
```

```python
from scipy.stats import ttest_1samp
```

```python
stats,p_value=ttest_1samp(sample_df['avgMonthlyHours'],proposed_mean)
```

```python
print(p_value)
```

```python
alpha=0.05
if p_value<alpha:
  print('There is a significant difference in between average monthly working hours')
else:
  print('There is no significant difference in average monthly working hours')
```

```python
total=len(sample1['left'])
total
```

```python
population_percentage=0.03
```

```python
from statsmodels.stats.proportion import proportions_ztest
```

```python
stats,p_value=proportions_ztest(count,total,population_percentage)
```

```python
print(p_value)
```

```python
if p_value>0.05:
  print('accept the null hypothesis')
else:
  print('reject the null hypothesis')
```
